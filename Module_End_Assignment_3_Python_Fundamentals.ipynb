# Step 2: Add More Feedbacks
● Ask the user to enter how many more feedbacks they want to add.

● For each feedback, collect the following inputs from the user:

○ Name

○ Written Feedback (text)

○ Rating (1–5)

● Automatically increment S_No starting from 11 onward.

● Append all new data into the feedback_data dictionary.

```python
f=int(input("How many more feedbacks you want to add ?"))
for i in range(f):
  Name=input("Please enter your name:  ")
  Feedback=input("Please enter your feedback:  ")
  Rating=int(input("Please enter your rating(1 to 5):  "))
  while Rating < 1 or Rating > 5:
    print("Not a correct rating, Please enter again:")
    Rating = int(input("Please enter your rating (1 to 5): "))
  S_No=len(feedback_data['S_No'])+1
  feedback_data['S_No'].append(S_No)
  feedback_data['Name'].append(Name)
  feedback_data['Feedback'].append(Feedback)
  feedback_data['Rating'].append(Rating)
  print(feedback_data)
```

# Step 3: Text Cleaning
Clean all feedback entries as follows:

● Remove punctuation (.,  ,,  !,  ?)

● Replace multiple spaces with a single space

● Remove leading and trailing spaces

● Convert all text to lowercase


```python
for i in range(len(feedback_data['Feedback'])):
  a=feedback_data['Feedback'][i]
  a=a.replace(".","")
  a=a.replace(",","")
  a=a.replace("!","")
  a=a.replace("?","")
  a=" ".join(a.split())
  a=a.lower()
  feedback_data['Feedback'][i]=a
print(feedback_data['Feedback'])
```

For loop is used to repeate the process

len() function counts the no of comments and range() function take index numbers

a,the variable used to store the value and replace() function is used in 4 lines to change the .,? etc

# Step 4: Word Count Insights
Create a function count_word_in_feedbacks(word) that:

● Takes a word as input.

● Returns how many feedbacks contain that word (case-insensitive match).

Use this function to print:

● Number of feedbacks containing "good"

● Number of feedbacks containing "poor"

● Number of feedbacks containing "excellent"


```python
def count_word_in_feedbacks(word):
  count=0
  for i  in feedback_data['Feedback']:
    if word in i.split():
      count+=1
  return count
print("No of feedbacks containing good:" ,count_word_in_feedbacks("good"))
print("No of feedbacks containing poor:" ,count_word_in_feedbacks("poor"))
print("No of feedbacks containing excellent:" ,count_word_in_feedbacks("excellent"))
```
